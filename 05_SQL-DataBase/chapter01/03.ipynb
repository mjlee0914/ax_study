### DDL
1. DDL(Data Definition Language): 데이터 정의어, **삭제시 rollback X**
    - CREATE [table, idndex, type...]
    - ALTER:
      - 컬럼 추가: alter table 테이블명 add column 컬럼명 자료형 [제약조건];
      - 컬럼 삭제: alter table 테이블명 drop column 컬럼명;
      - 컬럼 수정: alter table 테이블명 rename column 기존컬럼명 to 변경컬럼명
      - 컬럼 dtype 변경: alter table 테이블명 alter column 컬럼명 type 자료형; (구조변경, 데이터가 없는 경우)
                                                              -  자료형 using (컬럼명::자료형); (데이터가 존재하고 형변환이 필요한 경우)
      - 컬럼 제약조건 변경: alter table 테이블명 alter column 컬럼명 set 제약조건/default 
      - 컬럼 제약조건 삭제: alter table 테이블명 alter column 컬럼명 drop 제약조건/default
      - 컬럼 제약조건 추가: 
        - ex) alter table 테이블명 add constraint 제약조건구문이름 PRIMARY KEY(컬럼명, ...);
        - alter table 테이블명 add constraint 제약조건구문이름 FOREIGN KEY(현재 테이블 컬럼명) REFERENCES 참조할 테이블(컬럼명) ON DELETE CASCADE | SET NULL; 
        - ex) alter table 테이블명 add constraint 제약조건구문이름 unique(컬럼명, ...);
        - ex) alter table 테이블명 add constraint 제약조건구문이름 check(조건식);
    - DROP: [table, idndex, type...]
      - drop constraint 제약조건이름
    - TRUNCATE: 테이블 모든 데이터 삭제 (테이블의 구조는 유지)


# 테이블 간 관계
- 1:N 관계: 어떤 테이블 레코드 하나 -> 다른 테이블 여러 레코드 연결 ex) 회원 테이블 - 주문 테이블
- N:M (다대다) 관계: 양쪽 테이블 레코드가 서로 여러 개와 연결될 수 있음, 중간 테이블 거쳐야 함
- 1:1 관계: 한 테이블의 레코드 하나 -> 다른 테이블 레코드 하나와 대응 ex) 회원 테이블 - 회원 상세 정보 테이블



- 상품 -> 주문상세 꺾쇠 반대로 되어있음
- 상품 이름도 주문상세 테이블에 있어야 함
- 반정규화 필요 상황: 상품변경시점이 주문시점보다 이후에 일어날 경우 주문상세 내용이 변경되면 안됨

### DML Data Manipulation Language
1. INSERT

`INSERT INTO 테이블명 [(컬럼명, ..., ....)] VALUES (값, ..., ....);`
- 모든 컬럼명에 순서대로 추가할 경우는 컬럼명 생략 가능 (대괄호는 생략 가능)

2. UPDATE

`UPDATE 테이블명 SET 컬럼명 = 변경할 값, 컬럼명2 = 변경할 값2, ...`

3. DELETE

`DELETE 테이블명`

`DELETE FROM 테이블명 WHERE 조건`
