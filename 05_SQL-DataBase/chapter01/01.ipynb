데이터베이스 (DataBase Management System)
- 데이터를 관리하는 소프트웨어 혹은 데이터의 집합
- 데이터: 처리되지 않은 값(원시값). 이름, 나이, 구매금액 ...
- 정보: 데이터에 **특정 맥락의 의미**를 부여

SQL (Structured Query Language)
- query: 데이터베이스에 모이는 질문, 질의
- 작성 규칙: SQL

RDBMS(Relational DataBase System)
- table간의 관계를 명시하고, 관계를 바탕으로 데이터가 설계되는 형태의 데이터
- RDBMS 종류
  - MySQL
  - Oracle
  - MSSQL
  - Postgres -> vector DB 확장기능 "RAG"

- table: 데이터를 저장하기 위한 공간- 행과 열로 구성된 2차원 구조
  - 행: (Row/Record)
  - 열: (Column/Field/속성 ...)
  - 셀: 각각의 데이터 (주문번호, 주문상품 ....)

- schema: 테이블 구조와 제약사항을 정의한 명세서
