#### 4.1 데이터 로드 및 결측치, 이상치 처리

타이타닉(Titanic) 탑승객 데이터 로드 하기

- `df.head()`: 앞쪽 5개 조회
- `df.tail()`: 뒤쪽 5개 조회

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("titanic")

df.iloc[:5]
df.head()
df.tail()
```

데이터 요약 및 통계 확인 하기

```python
# 컬럼 목록 확인
df.columns
```

```python
df.info() # 결측치 확인
```

```python
df.describe() # 이상치 확인
```

결측치 파악하기

```python
df.isna().sum() # 결측치 갯수 파악
```

숫자형 데이터(나이): 전체 탑승객의 '평균(mean)'으로 빈칸 채우기

```python
mean_age = df['age'].mean() # mean() 기본값 axis=0
print("평균 나이:", mean_age)

df['age'] = df['age'].fillna(mean_age)

df.isna().sum()
```

문자/범주형 데이터(탑승 항구): 가장 많이 탑승한 곳인 '최빈값(mode)'으로 빈칸 채우기

```python
mode_embarked = df['embark_town'].mode()[0] # 최빈값 하나일 경우 mode(), 여러개일 경우 mode()[0]

mode_embarked
```

```python
df['embark_town'] = df['embark_town'].fillna(mode_embarked) # 최빈값으로 embark_town fill up
df['embarked'] = df['embark_town'].map(lambda x: x[0].upper()) #embark는 embark_town의 첫번째 대문자이므로, 업데이트된 embark_town을 고려하여 map(lambada)

df['embarked']
df['embark_town']
df.isna().sum()
```

특정 컬럼(deck) 을 '기준'으로 좌우(열 방향) 데이터를 채우기

```python
df['deck'] = df['deck'].ffill()
df.isna().sum()
```

```python
df.head() # deck 0번째 데이터 NaN

df['deck'] = df['deck'].bfill() #뒤쪽값으로 전부 refill 

df.head() 
```

너무 비싼 요금(이상치) 걸러내기

```python
fare_mean = df['fare'].mean()
fare_median = df['fare'].median()
fare_max = df['fare'].max()
print("요금 평균:", fare_mean, "요금 중앙값:", fare_median, "요금 최댓값:", int(fare_max))


# 그래프 그려서 요금 확인

plt.figure(figsize=(5,5))
plt.hist(df['fare'], bins=10)

plt.show() # x축(fare) 500에 미세하게 있음 / fare 300 = first class => 300이상의 값은 outlier 처리
```

```python
df = df[df['fare'] < 300 ]
```

결측치/이상치 보완 결과 확인

```python
df.isna().sum()
df[df['fare'] > 300].count() # 300을 초과하는 이상치는 배제됨
```

#### 4.2 중복 데이터 처리하기

중복 데이터 확인하기

```python
df.duplicated()
df.duplicated().sum()
```

중복 데이터 제거하기

`drop_duplicates()`: 중복 행 삭제

```python
df.drop_duplicates()[:100]
df = df.drop_duplicates().reset_index(drop=True) # 중복 행 삭제 + 인덱스 초기화
print("중복 데이터 수:", df.duplicated().sum())
```

#### 4.3 그래프로 시각화하기

내 컴퓨터에 맞는 한글 폰트 설정하기

```python
import platform
from matplotlib import rc

os_name = platform.system()

if os_name == "Windows":
    rc("font", family="Malgun Gothic")
else:
     rc("font", family="Apple Gothic")
```

도화지(Figure)와 3개의 액자(Axes) 준비 + 그래프로 분석하기

```python
fig, axes = plt.subplots(1, 3, figsize= (18, 5))

fig.suptitle("타이타닉 탑승객 데이터 분석", fontsize=16)

# [히스토그램] 탑승객 나이 분포
axes[0].hist(df['age'], bins=20, color="pink", edgecolor='black')
axes[0].set_title("<히스토그램>\n탑승객의 나이분포")
axes[0].set_xlabel("나이")
axes[0].set_ylabel("승객수")

# [박스 플롯] 티켓 등급에 따른 요금 이상치 확인
sns.boxplot(x='class', y='fare', data=df, ax=axes[1], color="skyblue")
axes[1].set_title("<박스플롯> 티켓 등급별 요금 분포")
axes[1].set_xlabel("티켓 등급")
axes[1].set_ylabel("티켓 요금")

# [산점도] 나이와 클리닝된 요금의 관계

axes[2].set_title("<산점도>\n나이 vs 티켓 요금(빨간색:생존 | 파란색:사망)")
scatter = axes[2].scatter(
    df['age'],
    df['fare'],
    c=df['survived'], 
    cmap='coolwarm', # 생존 1=빨강 ,사망 0=파랑
    s=df['fare'] / 2, #산점도 원 사이즈 지정
    alpha=0.7 # alpha=1 불투명, 0에 가까울수록 투명
)

axes[2].set_xlabel("나이")
axes[2].set_ylabel("티켓 요금")
fig.colorbar(scatter, ax=axes[2], label="생존여부") #컬러맵 추가 -> 파란색일수록 사망 
```
