# 1장 1강: 머신러닝의 이해

### 머신러닝
1. 데이터 사이언스의 한 분야
2. 데이터에 기반한 규칙을 발견하는 분야
    - 알고리즘: 예측 / 분류
3. 학습 -> 규칙

    - 학습 방법:
            
            - (1) 지도학습(Supervised Learning) <문제(feature 혹은 독립변수) + 답(label 혹은 종속변수)>
            - (2) 비지도학습(Unsupervised Learning) <문제/특성 => 규칙 찾기>
                    - ex. K-Means => 클러스터링, 군집
            - (3) 준지도학습 <지도학습 + 비지도학습>
            - (4) 강화학습 : 보상 알고리즘


4. 규칙을 통해 해결하고자 하는 문제영역
    - (1) 회귀 (Regression): 수치 예측
    - (2) 분류 (Classification)


5. AI
    - Strong AI
    - Weak AI: 특정 분야에 한정된 AI


6. 알고리즘
    - (1) Regression - 선형 회귀 
    - (2) ANN (Artificial Neural Network): 인공 신경망(퍼셉트론) -> DEEP LEARNING -> DNN(Deep Neural Network) -> CNN, RNN/LSTM/GRU, GAN
    - (3) KNN - 일상적 문제 해결 (인공 신경망 개발 빙하기)
    - (4) SVM
    - (5) Decision Tree
    - (6) Ensemble
