DataFrame

- 2차원(행, 열) 구조체
- 행 인덱스 + 열 이름 column

```python
data = {
    "이름": ["김철수", "이영희", "박민수"],  
    "부서": ["개발", "기획", "개발"],
    "급여": [4500, 5200, "4800"]
}

df = pd.DataFrame(data)
print(df) # 행 인덱스 0부터 시작 + column은 이름, 부서, 급여
```

#### 2.3 데이터 변경 및 인덱스 제어(set_index, reset_index)

```python
# df['급여'], type(df['급여'])

# 보너스 column 추가
df['보너스'] = df['급여'] * 0.1
df
```

```python
# 인덱스를 변경하는 경우
df_indexed = df.set_index('이름')
df_indexed
```

```python
# 변경한 인덱스를 다시 숫자형 인덱스로 초기화
df_restored = df_indexed.reset_index('이름')
df_restored # new dataframe
print(df_restored)
print(df is df_restored) # False
```

#### 2.4 Pandas Series & DataFrame 브로드캐스팅 연산

- axis: 0: -> 행 index 기준
- axis: 1: -> 열 index 기준


```python
df = pd.DataFrame({
    '중간고사': [80, 95, 70],
    '기말고사': [85, 90, 75] #인덱스 총 3개 (0, 1, 2)
}, index=['김철수', '이영희', '박민수']) # index=로 인덱스 설정 DataFrame({data dictionary}, index=[value 개수])

# df = df.set_index("김철수", "이영희", "박민수") DOESNT WORK

df
```

```python
# 각 시험별 (column에 대한) 평균 series 계산 -> axis=0. 행 기준(인덱스 기준)
score_means = df.mean(axis=0)
score_means
```

```python
means_per_student = df.mean(axis=1)
means_per_student
```

```python
score_gap = df.sub(score_means, axis=1) # 학생 실제 점수 - 시험 평균
score_gap
```
