### 3. Pandas 정밀 조회 및 결측치 제어

#### 3.1  loc와 iloc를 활용한 행·열 선택 (check location)

loc
- `.loc['행 label', [column label]]`: 행의 이름으로 검색(선택)
    - `.loc['행 label']`: 행만 검색 가능
    - `.loc['행 label', 'column label']` => numpy.자료형(검색 결과값)
    - `.loc[시작행:종료행, [컬럼1, 컬럼2, ...]]`
    - `'loc[..., 시작컬럼:종료컬럼]`

iloc
- `iloc[행 위치, 열 위치]`: 숫자 인덱스로 검색(선택)
    - loc와 사용 방법은 동일하지만 행 라벨이 아닌 **인덱스 번호**
    - 범위를 설정할 경우 종료 번호는 미만
    - ex) iloc[0:2, 0:2] -> 인덱스 0-1까지, 컬럼 0-1까지 ==> **숫자는 대부분 '미만'**

```python
import pandas as pd

df = pd.DataFrame({
    '나이': [25, 30, 35, 40],
    '점수': [80, 95, 70, 85],
    '등급': ['B', 'A', 'C', 'B']
}, index = ['user1', 'user2', 'user3', 'user4'])

df
```

```python
df.loc['user1'], type(df.loc['user1'])
```

```python
df.loc['user1', '점수']
```

```python
df.loc['user1':'user4', ['점수', '등급']] # user1부터 user4까지 점수와 등급 검색
```

```python
df.loc['user1':'user2', '나이':'등급'] # user1부터 user2까지 나이부터 등급까지 검색
```

```python
df.iloc[0:2, 1:3] # iloc는 0:1까지, 1:2까지
```

#### 3.3 결측치(Missing Data) 7대 처리 기법
3.3.1 결측치 탐지 및 현황 요약 (Detection)

- 결측치 - np.nan, pd.NA, NONE


[결측치 확인]
- `df.info()` 확인
- `df.isna()`, `df.isnull()`
- `df.isna().sum()`: 컬럼별 결측치 총 개수
    - 기본값: `df.isna().sum(axis=0)` , 행 기준

```python
import numpy as np

raw_df = pd.DataFrame({
    'A': [1, np.nan, 3],
    'B': [pd.NA, 5, 6],
    'C': [7, 8, 9]
    })

print(raw_df)
print(raw_df.info()) # 결측치 확인
```

```python
print(True + True) # 2
print(False + 1) # 1
```

```python
print(raw_df.isna())
print("=" * 30)
print(raw_df.isnull())
print("=" * 30)
print(raw_df.isna().sum()) # 0 =>결측치 없음, 행 기준
print(raw_df.isna().sum(axis=1)) # 열 기준
```

3.3.3 통계적 대푯값 대치 (Imputation)

대푯값 -> 데이터 손실 방지
- 평균값 : `mean()`
- 중앙값 : `median()` --> 극단치(이상치)의 영향을 덜 받음

대치
- `채워질data.fillna(채울data)`

```python
df_age = pd.DataFrame({
    '나이': [20, 25, None, 30, 95] # 극단치 95
    })

print(df_age)

median_val = df_age['나이'].median()
print("중앙값:", median_val) # 중앙값 27.5

df_age['나이'] = df_age['나이'].fillna(median_val)
print(df_age) # 2행 결측값이 중앙값으로 채워짐
```

3.3.4 시계열 직전/직후 값 대치 (Forward / Backward Fill)

유사한 형식의 데이터가 근처에 있을 확률이 높으므로 직전 또는 직후의 유효한 값으로 결측치 대체

- `ffill()` : 앞쪽에 있는 값으로 대체
- `bfill()` : 뒤쪽에 있는 값으로 대체

```python
ts_df = pd.DataFrame({'온도': [18.2, None, None, 21.0]})
print(ts_df)

print("=" * 40)
# forward fill
print(ts_df.ffill())

print("=" * 40)
# backward fill
print(ts_df.bfill())
```

3.3.5 선형 보간법 (Linear Interpolation)

규칙이 있는 데이터일 경우, 유효 데이터의 변화율 계산하여 그에 따른 일정한 간격에 따라 결측치를 채움
- `df.interpolate(method='linear')`

```python
speed_df = pd.DataFrame({'속도': [10.0, None, None, 40.0]})

speed_df['속도'] = speed_df['속도'].interpolate(method='linear')

print(speed_df)
```

3.3.6 이상 기호 치환 및 수치형 변환 (Replace & Type Casting)

- `.replace([이상한 값1, 이상한 값2, ...], pandas.NA)`: 이상한 데이터를 교체할 때, 이상한 값을 결측치(pandas.NA)로 변환

- `pandas.to_numeric(...)`: 숫자 자료형으로 변환

```python
import pandas as pd
survey_df = pd.DataFrame({'점수': [85, -999, '?', 95]})
survey_df.info()
```

```python
survey_df.describe()
```

```python
survey_df['점수'] = survey_df['점수'].replace([-999, '?'], pd.NA)
print(survey_df.info())
```

```python
survey_df['점수'] = pd.to_numeric(survey_df['점수'], errors='coerce')
survey_df.info()
```

```python
survey_df.describe()
```

#### 3.4 그룹화 집계(groupby)와 .to_numpy() 상호 변환

- `.groupby(column)`
- `.mean(numeric_only=True)`
- `.to_numpy(dtype=numpy.자료형)`
    - to_다른 dataformat/type 가능

```python
import numpy as np


sales_df = pd.DataFrame({
    '지점': ['서울', '서울', '부산', '부산'],
    '매출': [100, 150, 80, 120]
})

sales_df
```

```python
grouped_df = sales_df.groupby('지점').mean(numeric_only=True) #그룹화 집계, numeric_only=True 필수

grouped_df
```

```python
numpied = sales_df['매출'].to_numpy(dtype=np.float64) # DataFrame -> Numpy 배열 추출

numpied
```
