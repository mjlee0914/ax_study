- React.js 기반 프레임워크

- Rust로 개발

JSX
- **JavaScript XML**
- A **JavaScript syntax extension** that lets us write **HTML-like syntax inside JavaScript**
  - ❌ not actual HTML
  - ✅ JSX is converted into JavaScript
- Mainly used in React to describe **what the UI should look like**
- Makes UI/component code easier to read and write

- **Process**
  ```text
  JSX / TSX
      ↓
  JSX transformer / transpiler
  (Babel, SWC, esbuild, etc.)
      ↓
  JavaScript
  ```

  - Babel is **one possible transpiler**, not always the one being used.
  - `.jsx` = JavaScript + JSX
  - `.tsx` = TypeScript + JSX

- **Component**
  - A reusable piece of UI, usually written as a JavaScript function
  - In JSX, custom components are written like **custom tags**
  - Component names start with a **capital letter**

  ```jsx
  function Header() {
    return <h1>Hello</h1>;
  }

  <Header />
  ```

- **Build / Bundling**
  - Components are JavaScript/TypeScript **modules**
  - The build tool follows their `import` / `export` dependencies
  - A bundler/build tool such as **Vite, Webpack, Turbopack, etc.** transforms, optimizes, and bundles the code

  ```text
  App.jsx
  ├── Header.jsx
  ├── Button.jsx
  └── style.css
          ↓
      build tool
          ↓
  build output
  ├── JavaScript chunks
  ├── CSS
  ├── HTML
  └── other assets
  ```

- So:
  ```text
  React components/modules
          ↓
  Build tool / Bundler
          ↓
  optimized JS + CSS + HTML/assets
          ↓
  Browser
  ```
