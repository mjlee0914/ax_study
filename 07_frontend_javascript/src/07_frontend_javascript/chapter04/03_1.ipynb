`NaN`

    -  `Number()` 감싸서 호출시 NaN이 나오면 true, 아니면 false
    -  문자열 숫자는 false로 인식해서 판별 어려움 -> 정확한 판별은 `typeof()` 사용

배열 Array in JS
- 배열 객체
- python list와 비슷하게 동작
- 어떤 type의 값도 지정 가능
- 데이터 군집 (컬렉션 중 하나)
- Array.length != Array index
- Array 생성자 함수를 통해 생성 
  - 매개변수 하나일 때 : 0 이상의 정수 => 배열 요소의 갯수
  - 두개 이상: 


````javascript
// 새 배열을 만드는 세가지 방법

// 배열 리터럴 표기법을 사용하여 'fruits' 배열을 만듭니다.
const fruits = ["Apple", 3, {name: "김철수"}, function abc() {}];
console.log(fruits.length); // 4
fruits[5] = "망고"; // 배열 인덱스 번호로 새로운 value 추가
console.log("length:", fruits.length); // 6. index 3= empty item
console.log(fruits)
console.log("인덱스 기준 길이:", fruits.length - 1);
delete fruits[1];
console.log(fruits)
// prototype: Array

// Array() 생성자를 사용하여 'fruits2' 배열을 만듭니다.
const fruits2 = new Array("Apple", "Banana");

// String.prototype.split()을 사용하여 'fruits3' 배열을 만듭니다.
const fruits3 = "Apple, Banana".split(", ");
````

- Array는 Object에 0, 1, 2, 3 을 준 것과 같음

```javascript
const fruitsObj = {0: "사과", 1: "포도", 2: "배", length: 3};
const fruitsArr = ["사과", "포도", "배"];
```

- `Array.prototype`을 기준으로 상속 여부 판단 
  - fruitsObj의 prototype = Object
  - fruitsArr의 prototype = Array

- `instanceOf()`: prototype chain 관계를 통해 객체 상속 관계 확인
  - `isArray()`로 짧게 활용 가능

Array 주요 메서드
- Create:
  - `.push()`: 요소 끝에 추가
  - `.unshift()`: 요소 앞에 추가
  - `.splice(시작, 삭제갯수, ...추가요소)`: 지정된 위치에 있는 특정 요소를 추가/삭제
- Read:
  - `forEach(callback)`
  - `indexOf()`: 요소 위치 번호 (왼->오)- 중복값 있으면 가장 왼쪽에 있는 것부터 찾음
  - `lastIndexOf()`: 요소 위치 번호 (오->왼)
- Update:
  - `.map(callback)`: callback 함수대로 요소 변경
  - `.filter(callback)`: callback 함수대로 필터링
- Delete: 
  - `.pop()`: 끝 요소 꺼내기(반환값 내보내고 제거)
  - `.shift()`: 앞 요소 꺼내기(반환값 내보내고 제거)
  - `.splice()`



****


- Create, Delete -> 원본 데이터 변경
- Update -> 새로운 배열 객체 반환
