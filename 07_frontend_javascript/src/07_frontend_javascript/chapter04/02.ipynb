- 조건식: 참, 거짓을 판별하는 연산 식(비교, 논리연산이 주로 쓰인다.)

```javascript
if (조건식) {
    /// 조건식 참일때 실행되는 구간
}


if (조건식) {
    // 조건식 참일때 실행되는 구간
} else if (....) {
    ...
} else if (....) {

} else { // 모든 조건이 거짓일때 실행되는 구간
    ...
}
```


삼항 조건식
- `(조건식) ? 참일 때 실행되는 코드 값 : 거짓일 때 실행되는 코드나 값` (총 3항- `1항 ? 2항 : 3항`)

````javascript
let num = 10;

const result = num === 10 ? "숫자가 10입니다." : "숫자가 10이 아닙니다."; 
console.log(result);
````

````javascript
let greeting;

greeting = greeting || "기본값"; // greeting이 False이므로 "기본값"

console.log(greeting);
````

````javascript
let greeting = "값";

greeting = greeting || "기본값"; // greeting이 True이므로 "값"

console.log(greeting);
````

````javascript
let num = 0;

num = num || 10; // 0은 false이기 때문에 10이 num에 할당됨.

console.log(num); 
````

0을 null로 두고싶지 않을 때 쓰는 연산자 
- **`??` -> nullish 연산자 -> null, undefined 일 때만 판단**
- 요즘은 `||`보다 많이 사용

````javascript
let num = 0;
num = num ?? 10; // null로서가 아닌 0으로서 취급
console.log(num);
````

#### 6.2 선택문 (switch - case)

- `switch ~ case`: case 키워드에 매칭이 되면 그게 바로 코드 실행 시작점, break 만날 때까지 실행
- `break` : 실행흐름 중단
- `default` : case에 매칭된 키워드가 없음녀 마지막에 유입되는 구간

```javascript
switch (expression) {
    case caseExpression1:
        statements;
    case caseExpression2:
        statements;
        break;
    // …
    case caseExpressionN:
        statements;
    default:
        statements;
}
```

````javascript
const rank = 4;

switch(rank) {
    case 1: // rank가 1인지 체크
        console.log("금메달");    // 숫자 1일 때 코드 
    case 2:
        console.log("은메달");  // "은메달" 출력 후 break
        break;
    case 3:
        console.log("동메달");
    default:
        console.log("등수는 1, 2, 3 중에서 입력!");
}
````

- break 없을 때

- break 있을 때

- default 적용

````javascript
/*
let grade = 1~ 6;
1 = 1학년, 6 = 6학년, 1 ~ 6 사이 값이 아니면 "잘못 입력" 출력
*/
````

````javascript
const grade = 6;

switch(grade) {
    case 1: 
        console.log("1학년");  
        break;   
    case 2:
        console.log("2학년");  
        break;
    case 3:
        console.log("3학년");
        break;
    case 4:
        console.log("4학년");
        break;
    case 5:
        console.log("5학년");
        break;
    case 6:
        console.log("6학년");
        break;
    default:
        console.log("잘못 입력");
}
````

````javascript
const grade = 6;

for (let i = 1; i <= 6; i++) {
    if (grade === i) {
        console.log(`${i}학년`);
        break;
    }
    if (i === 6) {
        console.log("잘못 입력");
    }
}
````

````javascript
const grade = 6;
const gradeNum = [1, 2, 3, 4, 5, 6];

for (let i = 0; i < gradeNum.length; i++) {
    if (grade === gradeNum[i]) {
        console.log(`${gradeNum[i]}학년`);
        break;
    }

    if (i === gradeNum.length - 1) {
        console.log("잘못 입력");
    }
}
````

````javascript
// 2 ~ 9 단까지 중첩된 for문 이용하여 작성
// 단보다 곱하는 수가 클때만 출력
for (let i = 2; i < 10; i++) {
    console.log(`=====구구단 ${i}단=====`);
    for (let j = 1; j < 10; j++) { 
        if (i < j) continue;

        console.log(`${i} x ${j} = ${i * j}`);
    }
}
````

do-while 문 : 일단 한 번 실행하고 조건 비교하기

- while 문
```javascript
while(조건식) {
    // 조건식이 참일 때 반복되는 코드
    // 반복을 중단시키는 코드가 있어야 하므로 break와 함께 쓰인다
}
```

- do while 문
```javascript
do {
    // 조건식이 참일 때 반복되는 구간
} while (조건식);
```


- while vs. do while
  - do while: 조건식이 거짓이더라도 한번 실행을 보장

````javascript
// chrome devtool console에서 동작
while (true) {
    const menu = prompt("메뉴 선택");

    if (menu === "q") {
        console.log("중단합니다...");
        break;
    }
}
````

````javascript
let i = 0, total = 0;

while(i <= 100) {
    total += i;
    i++;
}
console.log("합계:", total);
````

````javascript
let i = 101, total = 0;

do { // i 는 101이지만 일단 do 한번 실행.
    total += i;
    i++;
} while (i <= 100);

console.log("total:", total, "i:", i);
````

### 7. 예외 처리 (try - catch - finally)

#### 7.1 try - catch - finally 문의 구조

````javascript
try {
    let str;
    const str2 = str.toUpperCase();
    console.log(str2, "대문자로 변경 완료");
}   catch (err) {
    console.log(err);
} finally {
    console.log("무조건 실행...");
}
````

- `console.error(err)` 출력하여 error 붉게 출력

````javascript
try {
    let str;
    const str2 = str.toUpperCase();
    console.log(str2, "대문자로 변경 완료");
}   catch (err) {
    console.error(err);
}
````

- `console.dir(err)`: 에러메시지 스택으로 정리 - 발생한 예외메시지의 종류 등

````javascript
try {
    let str;
    const str2 = str.toUpperCase();
    console.log(str2, "대문자로 변경 완료");
}   catch (err) {
    console.dir(err);
}
````

````javascript
try {
    let str;
    const str2 = str.toUpperCase();
    console.log(str2, "대문자로 변경 완료");
}   catch (err) {
    console.log("err.message:", err.message, "err.name:", err.name);
}
````
