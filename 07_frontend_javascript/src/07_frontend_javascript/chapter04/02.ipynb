### 2. 초기 자바스크립트의 변수 선언: var의 치명적인 결함

```javascript
var a = 100
```

-> 번역 과정 -> window 객체의 하위 속성으로 추가 (웹브라우저 기준) / global 객체의 하위 속성으로 추가 (Node.js 환경)




- 윈도우 객체 호출 먼저 하고 정의해도 정상 작동 
- 호이스팅: 윈도우 아래쪽에 정의된 변수/함수들이 윈도우 상위객체로 이동
  - 정의한 변수, 함수가 최상위 객체에 할당.
  - 함수 정의 위치 상관없이 호출이 가능

#### 2.1 var의 3가지 결함
- 호이스팅 (Hoisting)
- 선언 키워드 생략 허용 (암묵적 전역 변수): var(선언자) 생략하고 변수를 지역변수로 정의하면 전역변수가됨 (고립시켜야하는데 코드 전체에 적용됨)
- 함수 레벨 스코프 (Function-level Scope): 전역변수 오염 문제 발생 => 대신 `let` 사용

--> 3가지 결함을 해결하기 위해 `ECMAScript 6` 나타남
    - const
    - let

````javascript
var i = 10000;

for (var i = 0; i < 10; i++) {
    console.log(`for문에 정의된 i: ${i}`) // console.log happens before i++
}

console.log(i);
````

````javascript
var a = 9999;

function add() {
    a= 10;
    console.log("지역변수", a);
}

console.log(a);
````

#### 3.2 let과 const의 구분 및 권장 사항
- let (재할당 필요 대상)
- const (재할당 불가 대상)
- 블록 스코프와 상수의 할당 검증

자바스크립트에서 변수 선언시 const로 선언, 변경이 필요한 경우만 let을 사용

````javascript
let c = 100;
c = 200;
console.log(c)
````

````javascript
const d = 100;
d = 200;
````

### 4. 데이터 타입과 typeof를 활용한 타입 검증

#### 4.1 자바스크립트의 주요 데이터 타입
- Number (숫자형): 정수, 실수 구분 x
- String (문자열): '', ""
- Boolean (불리언): true/false
- null: 값이 없는(비어있는) 상태임을 표시하는 자료형
- undefined: 값이 대입되지 않고 이름표(변수)만 정의되었을 때 기본값

````javascript
const num1 = 100;
console.log(typeof num1);

const str1 = "ABC";
console.log(typeof(str1));

const bool1 = true;
console.log(typeof(bool1));
````

````javascript
let num2;
console.log(num2);
console.log(typeof(num2));
````

````javascript
let num3 = null;
console.log(typeof num3); // object- 초기 설계 오류
````

#### 4.3 문자열의 확장: 문자열 결합과 템플릿 리터럴 (ES6+)

고전적 방식: 더하기(+) 연산자로 연결하기

````javascript
let name = "kim";
let greeting = "hello";

let text1 = name + greeting;
console.log(text1)
````

현대 표준 방식: 템플릿 리터럴 (Template Literals)

````javascript
const hello1 = '안녕하세요.\n반갑습니다';
console.log(hello1);
````

````javascript
const tpl = `${num1} + ${num2} = ${num1 + num2}`;
console.log(tpl);
````

````javascript
// 템플릿 함수
function func(...args) {
    console.log("args:", args);
}

console.log(func(tpl));
````
