What is Object?
- comprise properties and methods
  - properties
    - characteristics of an object
      - ex) key, value
  - methods
    - the value => functions/actions
      - a skill that an object has?
- group related data together
- split code into logical pieces
- key:value pair
  - key - string or symbol
  - value can be any data type
    - primitive value : str, int, ...
    - object value: not primitive value

````javascript
// key as str
const obj = {
    name: "Myungji",
    123: "hello"
};
````

````javascript
// key as Symbol
const id = Symbol("id");

const user = {
    [id]: 123
};
````

속성(property)
1. 조회   → obj.key (변수명 규칙 따를 경우) / obj["key"] (변수명 규칙 따르지 않을 경우)
2. 추가   → 없는 속성에 값 대입
3. 수정   → 있는 속성에 값 대입
4. 삭제   → delete obj.key / delete obj["key"]

일급 객체: 함수를 일반 값처럼 자유롭게 사용할 수 있음
- 숫자나 문자열처럼, 함수도 변수에 저장할 수 있다.
- 함수는 배열·객체에 넣을 수 있다.
- 함수를 다른 함수에 전달할 수 있다.
- 함수가 함수를 결과로 돌려줄 수도 있다.
```javascript
// 1. 변수에 담기 
const sayHello = function () {
  console.log("Hello!");
};

// 2. 배열에 넣기
const actions = [
  () => console.log("저장"),
  () => console.log("삭제")
];

// 3. 객체의 값(메서드)으로 넣기
const person = {
  name: "mj",
  introduce: function () {
    console.log(`저는 ${this.name}입니다.`);
  }
};

// 4. 다른 함수의 인자로 전달하기
function run(action) {
  action();
}

run(sayHello); // Hello!
```

#### EC와 Call Stack 한눈에 보기
```javascript
function inner() {
  console.log("inner 실행");
}

function outer() {
  inner();
}

outer();
```
① 프로그램 시작

| Call Stack |
| --- |
| **Global EC**<br>전역 변수·함수 |

② 함수 호출

| Call Stack |
| --- |
| **outer EC** |
| **Global EC**<br>전역 변수·함수 |

③ 함수 안의 함수 호출

| Call Stack |
| --- |
| **inner EC**<br>현재 실행 중 |
| **outer EC**<br>실행 대기 |
| **Global EC**<br>실행 대기 |

④ 호출이 끝남

| Call Stack |
| --- |
| **Global EC**<br>다음 코드 실행 |


**실행 컨텍스트(EC)**  
코드 실행에 필요한 변수, 매개변수, `this` 등을 담는 작업 공간

**콜 스택(Call Stack)**  
EC를 호출 순서대로 쌓고, 끝난 함수부터 위에서 제거하는 공간

closure: 함수가 만들어질 당시의 바깥 스코프 변수를 기억해서, 바깥 함수의 실행이 끝난 뒤에도 그 변수에 접근할 수 있는 기능

    - inner가 이미 끝난 outer의 num2: 10을 계속 기억 가능

#### 1.2 프로퍼티 읽기와 조작 (추가, 수정, 삭제)
property(속성) 접근법
- 마침표(.) 연산자: 속성 이름이 변수 이름 규칙에 해당하는 속성만 접근 가능
  - **변수명 규칙**
    - `알파벳(대소문자)`, `숫자`, `_$` 로만 구성
    - 숫자는 앞에 올 수 없음
    - 예약어(if, throw, ...)는 사용불가
  - **변수명 관례**
    - CamelCase
    - 상수(const) 있는 경우 : 대문자_대문자 ex) `NO_OF_STDUENT`
  - **함수명 관례**
    - 변수명 관례와 동일
    - 함수 중에서 생성자로 사용되는 함수는 앞글자를 대문자로 쓰는 것이 관례
      - ex) OrderItem
- 대괄호([]) 연산자: 변수명 규칙에 해당하지 않는 속성에 접근하는 경우
- 없는 프로퍼티 접근

ex) 
```javascript
const person = {
    my-name = "john"
}

person['my-name'] = "mj" // -는 변수명에 사용 안하므로 []로 접근
```

````javascript
console.log(person.age);
console.log(person['age'], person['name']);
````

- 변수명 규칙에 해당하지 않는 속성은 마침표 연산자로 접근 불가

````javascript
console.log(person.'student-no');
````

- 대신 대괄호 연산자 사용 가능

````javascript
console.log(person['student-no']);
````

- 동적으로 객체에 속성 추가
  - typescript 제한 걸리므로 web에서 확인

````javascript
person.phone = '010-1111-2222';

console.log(person)
````
