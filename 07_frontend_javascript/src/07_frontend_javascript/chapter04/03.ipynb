#### 1.2 프로퍼티 읽기와 조작 (추가, 수정, 삭제)
- 마침표(.) 연산자: 속성 이름이 변수 이름 규칙에 해당하는 속성만 접근 가능
  - **변수명 규칙**
    - `알파벳(대소문자)`, `숫자`, `_$` 로만 구성
    - 숫자는 앞에 올 수 없음
    - 예약어(if, throw, ...)는 사용불가
  - **변수명 관례**
    - CamelCase
    - 상수(const) 있는 경우 : 대문자_대문자 ex) `NO_OF_STDUENT`
  - **함수명 관례**
    - 변수명 관례와 동일
    - 함수 중에서 생성자로 사용되는 함수는 앞글자를 대문자로 쓰는 것이 관례
      - ex) OrderItem
- 대괄호([]) 연산자: 변수명 규칙에 해당하지 않는 속성에 접근하는 경우
- 없는 프로퍼티 접근

```javascript
console.log(person.age);
console.log(person['age'], person['name']);
```

- 변수명 규칙에 해당하지 않는 속성은 마침표 연산자로 접근 불가

```javascript
console.log(person.'student-no');
```

- 대신 대괄호 연산자 사용 가능

```javascript
console.log(person['student-no']);
```

- 동적으로 객체에 속성 추가
  - typescript 제한 걸리므로 web에서 확인

```javascript
person.phone = '010-1111-2222';

console.log(person)
```

#### 1.4 메서드 (Method)와 this
- 메서드
- this 키워드

```javascript
const object1 = {
    user: "mj",
    nationality: "South Korea",
    profession: "Software Enginneer",
    myBio() {
        console.log(`My name is ${this.user}. I'm a ${this.profession} from ${this.nationality}.`) // this. referes to the object itself
    }
}
console.log(object1.user);
console.log(object1.nationality); 
console.log(object1.profession); 
console.log(object1.myBio()); 
```
