#### 호스트 객체
- 런타임마다 다르게 구성되어 있는 특화 객체
1. 웹브라우저 객체
   - 브라우저의 기능과 밀접하게 관련

2. 윈도우 객체
   - 웹 브라우저 열 때마다 실행되는 객체, 가장 상위 객체
  1. location 객체: 주소창과 관련된 정보, 기능 제공
       - assign(이동할 주소): 주소로 이동
       - replace(이동할 주소): 주소로 이동 + 방문기록 남기지 않음
         - ex) 상품 장바구니 담고 뒤로가기 누르면 장바구니에 2번 등록되는 상황 등을 대비, 방문기록 남기지 않음
       - reload(): 새로고침
       - 속성: 실시간성 -> 주소창 구성요소가 실시간성을 가지고 변경내용이 반영됨.
        ex) location.href : 속성을 변경하면 페이지 이동 

  2. history 객체: 방문기록과 관련된 정보, 기능 제공
  3. screen 객체: 브라우저 사용 화면 정보
  4. navigator 객체: 브라우저 OS 사용 환경, 기타 API ex) 지도 api, 장비 api 등 getUserAPI
  5. document 객체: HTML 문서와 관련된 정보, 기능 제공

Geolocation
- 위치에 대한 정보
- 위치기반서비스시 사용

````javascript
// chrome console
var options = {
  enableHighAccuracy: true,
  timeout: 5000,
  maximumAge: 0,
};

function success(pos) {
  var crd = pos.coords;

  console.log("Your current position is:");
  console.log(`Latitude : ${crd.latitude}`);
  console.log(`Longitude: ${crd.longitude}`);
  console.log(`More or less ${crd.accuracy} meters.`);
}

function error(err) {
  console.warn(`ERROR(${err.code}): ${err.message}`);
}

navigator.geolocation.getCurrentPosition(success, error, options);
````

Output:
```
Your current position is:
VM170:11 Latitude : 37.48707980486462
VM170:12 Longitude: 126.89405013650737
VM170:13 More or less 36 meters.
```

window 객체의 주요 프로퍼티 (프로퍼티 값은 객체 생략하고 입력해도 console에서 확인 가능)
- innerWidth, innerHeight: 스크롤 막대를 제외한 창 안쪽의 실제 표시 영역 너비와 높이입니다.

- pageYOffset: y축으로 스크롤 얼마나 움직였는지 거리 반환
- pageXOffset: x축으로 스크롤 얼마나 움직였는지 거리 반환
