#### DOM object
1. DOM tree
- 2진 트리 -> 빠른 요소 탐색 가능
- React 등장 이유 -> 새로운 요소가 추가되거나 변경시 DOM tree를 재구성 -> 연산을 많이해서 사이트 느려지고 무거워짐 . React는 가상의 DOM tree를 그림. 
- facebook처럼 자진변경이 있는 웹에서 강점을 보임

2. 요소(DOM) 선택 방법
   1. `document.getElementById("id_name")`
   2. `document.getElementsByClassName("class_name")`
   3. `document.getElementsByTagName("tag_name")`
   4. `document.getElementsByName("name")`
   5. `document.querySelector("css선택자")`: 1개만 선택
   6. `document.querySelectorAll("css선택자")`: 모두 선택
   7. 문서 전체(html) - window.document
        - head: `window.document.head`
        - body: `window.document.body`
   8. 양식(form) 이름 - frmNIDLogin
    - 양식 하위 입력 태그는 이름으로 바로 접근 ex) frmNIDLogin.id
   9. 요소 노드(상대적 접근)
    - parentElement: 부모 요소 
    - childElement: 자식 요소
    - firstElementChild: 첫번째 자식요소
    - lastElementChild: 마지막 자식요소
    - nextElementSibling: 현재 기준에서 다음 레벨의 형제요소 
    - previousElementSibling: 현재 기준 이전 레벨 형제요소

**데이터 속성(`data-...`)**
- `document.dataset`: key-value 객체
- 실시간성 보장: 속성 추가/제거시 바로 적용
- data- 는 빠짐
- 
- 데이터삭제: delete(객체삭제는 delete)

- 아래처럼 type의 값이 바로 바뀌어버림

- 일반 속성들은 기능과 관련되어 있을 수 있기 때문에 영향을 주지 않기 위해 정보성 데이터는 앞에 `data-` 등 붙이는 것이 관례
- 

```javascript
type, name, id, value
→ HTML이 원래 알고 있는 속성, 기능에 영향을 줌

data-price, data-name, data-order
→ 개발자가 만든 추가 정보, 단순 데이터 저장 용도

data-* = HTML 요소에 개발자가 몰래 붙여두는 메모칸
```

### example

```html
<!-- html -->
<html>
  <head>
    <title>getElementById 예제</title>
  </head>
  <body>
    <p id="para">어떤 글</p>
    <button onclick="changeColor('blue');">blue</button>
    <button onclick="changeColor('red');">red</button>
  </body>
</html>
```

```javascript
// javascript
function changeColor(newColor) {
  var elem = document.getElementById("para");
  elem.style.color = newColor;
}
```
