TypeScript
- Superset : 자바스크립트 문법 포함 + 확장 형태
- 정적 타입: **코드 실행 전** (컴파일 타임) 오타/잘못된 타입 전달 등 버그 미리 잡아냄.
- 일반 언어의 특징 도입 (Java, C#) 
  - ex) TypeScript에는 class가 존재, 일반언어에서 쓰고 있는 spec들이 존재함.
- runtime에 타입 소멸 (Transpile process): JS engine 은 타입스크립트 문법을 인식 못하므로, 컴파일 과정에서 모든 타입 코드를 지우고 순수한 JS 문법으로 변환되어 실행됨.
  - 변환도중 타입체크시 이상 있으면 JS 문법으로 변환되지 않음
