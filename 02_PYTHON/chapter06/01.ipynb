- \_\_name\_\_ : 모듈 이름

## 2. 디렉터리형 패키지(Package)

- 패키지는 모듈을 모아놓은 디렉토리
- `from 패키지명.하위패키지 import ...`

- 전체 패키지 경로: 절대 경로 접근 ex) chapter06.mod1
- 상대 경로 접근법
    - `(..)`: 바로 위 패키지
    - `(.)`: 현재 모듈이 있는 패키지  

## 3. import, from, as  키워드

`import 모듈명`
- 모듈 불러오기
- 불러온 모듈에서 마침표(.)를 통해서 접근 = '모듈.'

`from 모듈명/패키지명 import 모듈에 있는 함수/변수/클래스/.../모듈`
- 호출 시 앞에 모듈. 붙이지 않고 함수 이름으로 바로 동작 가능

`from 모듈명 import *`
- 모듈의 전체 함수, 변수, 클래스를 불러온다.

`import 모듈명 as 키워드` : as(alias)
- 모듈명을 키워드로 불러오기 

```python
import mod1

mod1.add(10, 20)
mod1.VERSION
```

```python
import mod1 as m
m.add(1, 2)
```

```python
from mod1 import add, minus, VERSION
print(add(1, 2))
print(minus(5, 1))
print(VERSION)
```

```python
from mod1 import *
print(add(1, 2))
print(minus(5, 1))
print(VERSION)
```

#### \_\_init\_\_.py

- 현재 디렉토리가 패키지임을 알려주는 파일 (디렉토리 안에 생성되는 파일임)
- 오류 방지를 위해 패키지를 만들 때 해당 파일도 만들어주는 것이 좋음
- 하위 모듈, 하위 패키지에서 공통적으로 처리해야 하는 코드를 작성
- 모듈을 불러올 때 최초 한 번만 실행
- 패키지 내에서 사용하는 공통적인 함수, 변수, 클래스를 정의
- `from 패키지 import *` : `__all___ = [....]` : 에 정의된 모듈만 import 한다

```python
from game import lol
```

```python
import game #__init__.py 가 있으면 import로 파일 접근 가능
game.common_func()
```

```python
from game import *
print(game.lol.LOL)
print(game.wow.WOW)
```
