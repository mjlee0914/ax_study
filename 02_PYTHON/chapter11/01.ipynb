# 동기(Sync) vs 비동기(Async)

- 비동기 제어권: 작업을 시작한 뒤, 그 작업이 기다리는 동안 제어권을 다른 작업에게 넘긴다.
- `await`
- `asyncio.sleep()`
- `asyncio.gather()`



** Generator = 내가 next()로 제어,
Async = await로 제어권을 양보하고 Event Loop가 재개를 관리




```python
# 동기식 함수 (서브 루틴)
import time

def work(title, seconds):
    time.sleep(seconds) #지연
    print(title, "완료!")

start = time.perf_counter() # 작업 시작 시간

work("A작업", 2) # blokcing 
work("B작업", 3) # blocking ==> a,b 끝나야 c 작업 실행
work("C작업", 1)

end = time.perf_counter() # 작업 종료 시간

print("걸린 시간: ", end - start)
```

```python
# 비동기 함수 Co-routine
import asyncio
import time

async def async_work(title, seconds): # coroutine 정의
    await asyncio.sleep(seconds) # 기다리는 동안 제어권 양보
    print(title, "완료!")



async def main():
    start = time.perf_counter()

    await asyncio.gather( 
        async_work("A작업", 2),
        async_work("B작업", 3), #다른 작업 실행, 완료시 coroutine 자동 실행 
        async_work("C작업", 1)
    )

    end = time.perf_counter()
    print("걸린 시간:", end - start)


await main()
```
