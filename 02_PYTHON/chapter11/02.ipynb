```python
scores = [45, 80, 55, 90, 30]
result = list(filter(lambda x: x >= 60, scores))
print(result)
```

```python
scores = [45, 80, 55, 90, 30]
result = [x for x in scores if x >= 60]
print(result)
```

# 2. 실행 환경에 따른 가동 방식

- 주피터 노트북엥는 이벤트 루프가 가동 중 -> await 함수명()
- 파이썬 파일(py)를 실행할 때는 이벤트 루프를 생성해야 정상 동작 `asyncio.run(await함수)`


```python
import asyncio

async def work():
    await asyncio.sleep(2) #
    print("작업 완료")
```

```python
await work()
```
