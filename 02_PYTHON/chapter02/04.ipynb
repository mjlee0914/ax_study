## 1. 1세대: % 연산자 방식

- %s : string 문자열 대체
    - %숫자s -> 최소 숫자만큼 출력 공간 확보, 남는 공간은 공백으로 채움
- %d : integer 정수 대체
    - %숫자d ex) %05d = 5자리만큼 출력 width 확보, 남는 공간은 0으로 채움
    - 소수점은 날라감
- %f : float 실수 대체
    - %.2f : 소수점 2자리까지 표시
    - %10.2f: 전체 출력 폭을 10칸 확보. 소수점 자리는 2. % [전체 너비].[소수점 자리수] 타입 (dot 포함)

```python
message = "나의 이름은 %s이고, 나이는 %d입니다."

message % ("철수", 20) # "나의 이름은 철수이고, 나이는 20입니다."
```

```python
message = "주문번호: %05d" % 10
message # 주문번호: 00010, 5자리로 맞추고 남는 자리는 0으로 채움
```

```python
message = "숫자: %f" % 10.123456789
message # 숫자: 10.123457, 소수점 6자리까지 표시, 반올림 처리됨

message2 = "숫자: %.2f" % 10.123456789
message2 # 숫자: 10.12, 소수점 2자리까지 표시, 반올림 처리됨

message3 = "숫자: %010.2f" % 10.126456789
message3 # 숫자: 000010.13, 소수점 2자리까지 표시, 반올림 처리됨, 전체 자리수 10자리로 맞추고 남는 자리는 0으로 채움
```

```python
message = "숫자: %d" % 10.12345
message
```

## 2. 2세대: str.format() 메서드 방식

- 순서 지정 교체: {0} {1} {2}
- 단어 지정 교체: {교체 단어} {교체 단어}

```python
message = "내 이름은 {0}이고, 나이는 {1}살 입니다."
message.format("철수", 20)
```

```python
message = "내 이름은 {name}이고, 나이는 {age}세 입니다."

message.format(name = "철수", age=20)
```

```python
message = "내 이름은 {name}이고, 나이는 {0}세 입니다."
message.format(20, name="철수")
```

## 3. 3세대: f-string 방식

1. 변수 지정
2. f"str{지정한변수}"

```python
name = "영희"
age = 20
message = f"내 이름은 {name}이고, 나이는 {age}세 입니다."
message
```

```python
price = 10000

message = f"가격: {price}" 
message2 = f"가격: {price:,}"

print(message, message2)
```

```python
f"가격: {price:>10,}원" #왼쪽 공백
```

```python
f"가격: {price:<10,}원" #오른쪽 공백
```

```python
# 디버깅 조절자 (=) : 변수의 이름과 현재 값을 동시에 출력
f"{price=}"
```
