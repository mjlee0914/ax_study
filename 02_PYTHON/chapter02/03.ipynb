# 2장 3강: 연산자

## 1. 항과 연산자(Operator)

- 항 : 연산에 사용되는 값, 2개 -> 이항, 3개 -> 3항, 1개 -> 단항
- 연산자: 연산에 사용되는 기호

## 2. 연산자의 종류

#### 대입 연산자(=)
x = 10
x = 10 + 20

대입 연산자는 연산의 우선 순위가 가장 낮다. 가장 마지막에 연산된다.

#### 산술 연산자 arithmetic operators
+, -, /, *
// 나눈 몫
% (modulo operator) 나눈 나머지 
**  ex) 2**3 = 2 *2 *2 

- 산술 연산자 우선순위: **, *, /, //, % > +, -
- 우선 순위 강제 적용 : ()


#### 비교 연산자
.>, <, >=, <=

== : 같다(동등성, 동일한 가치)

!= : 다르다

- 연산의 결과는 무조건 True/False
- 판별이 필요한 구문 (조건문, 반복문)


동등성 vs 동일성
- 동등성 : 동일한 가치를 비교 (==)
- 동일성 : 동일한 대상인지 비교 (is - 주소 비교)

```python
num1 = 100
num2 = 200

# 동등성 비교
num1 == num2  # False, 값 다름

# 동일성 비교
id(num1), id(num2) # 140711679905024, 140711679905344
print (num1 is num2) # False, 주소 다름

x = num1
id(x), id(num1) # 140711679905024, 140711679905024
print (x is num1) # True, 주소 같음
```

```python
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

mod_ints = my_int_1 % my_int_2
mod_floats = my_float_2 % my_float_1

print('Integar modulo: ', mod_ints)
print('Float modulo; ', mod_floats)
```

```python
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

floor_div_ints = my_int_1 // my_int_2
floor_div_floats = my_float_2 // my_float_1

print('Integar floor division: ', floor_div_ints)
print('Float floor division; ', floor_div_floats)
```

```python
my_int_1 = 2
my_int_2 = 3

my_float_1 = 2.3
my_float_2 = 3.4

exp_ints = my_int_1 ** my_int_2 #2 ** 3 = 2 * 2 * 2 
exp_floats = my_float_1 ** my_float_2 

print('Integer Exponentiation:', exp_ints) 
print('Float Exponentiation:',  exp_floats) 
```

#### 논리 연산자 (and, or, not)

and
- 각 항이 모두 True -> True, 하나라도 False -> False

or 
- 둘 중 하나만 True -> True, 모두 False -> False

not 
- True 조건 -> False 조건으로 반전, False 조건 -> True 조건으로 반전

:exclamation: 논리 연산자는 비교 연산자보다 우선 순위가 낮다

```python
# and
num1 = 10
num2 = 20

result1 = num1 < 20
result2 - num2 > 30

result1 and result2 # False, 두 조건 중 하나라도 거짓이면 False
```

```python
# or
result1 or result2 # True, 두 조건 중 하나만 참이어도 True
```

```python
# not
result3 = not result1 # False, result1이 True이므로 not 연산으로 False

result4 = not(result1 and result2) # True, result1 and result2가 False이므로 not 연산으로 True
```

## 3. 단락 회로 평가(Short-Circuit Evaluation)

```python
# or 연산 -> 왼쪽이 거짓이면 오른쪽 조건 확인+오른쪽 결과값 반환
result = False or 10 + 20 # 30, 첫 번째 조건이 False이므로(아직 결과가 결정되지 않았으므로) 두 번째 조건 확인하고 결과값을 반환
result1 = True or 10 + 20 # True, 첫 번째 조건이 True이므로 두 번째 조건은 확인하지 않음 

# and 연산 -> 왼쪽이 거짓이면 오른쪽 조건 확인하지 않음, 왼쪽이 참이면 오른쪽 결과값 평가+반환
result2 = False and 10 + 20 # False, 첫 번째 조건이 False이므로 오른쪽 조건은 확인하지 않음
result3 = True and 10 + 20 # 30, 첫 번째 조건이 True이므로 두 번째 조건 확인하고 결과값을 반환
```

```python
# or 연산을 이용한 기본값(없을 때 대체되는 값)
message = ""
result5 = message or "기본 메시지"

result5 # "기본 메시지", message가 비어있으므로 false, or은 왼쪽 false시 오른쪽 확인하여 오른쪽 값 반환
```

```python
# and 연산을 이용한 간단한 조건 처리
height = 165

result = height >= 160 and "크다"

result
```
