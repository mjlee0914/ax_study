### 1. 파이썬의 기본 자료형 (Primitive Data Types)

- 자료: 연산에 필요한 값(데이터)
- 자료형: 값의 유형(데이터의 종류)

#### 자료형
- 숫자형: 정수형(int, 10, 20, ...), 실수형(float, 10.123, ...)
- 논리형: bool 참(True), 거짓(False)
- 문자형: String, 문자/문장, '', "" 감싸서 선언(한줄), """...""" (여러줄 문장, 줄바꿈 유지 가능) 
- 값 없음: None

### bool()
- False : 0, "", [], None
- True : int/float, "str", [value]

```python
message = "1"
if message: # 문자열이 비어있지 않으면 True, 비어있으면 False
    print("yes")
```

```python
result = bool(message) # 문자열이 비어있지 않으면 True, 비어있으면 False
result, type(result)
```

### 2. 명시적 형변환(Explicit Type Conversion)

형변환 함수
- int() 
- float()
- bool()
- str()

:exclamation: can only concatenate the same type 


```python
num = "100"
type(num)

num + 200 # TypeError: can only concatenate SAME type (str, str) not str and int

x = int(num) + 200 # int() 함수를 사용하여 문자열을 정수로 변환 후 연산 가능
print(x) # 300
```

```python
print("현재 시간은 " + 10 + "시 입니다") # TypeError: can only concatenate SAME type (str, str) not str and int
print ("현재 시간은 " + str(10) + "시 입니다") # str() 함수를 사용하여 정수를 문자열로 변환 후 연산 가능
```

### 3. 묵시적 형변환(Implicit Type Conversion)

Type Promotion/Coercion 타입 승격
- 계산 중 정보 손실되지 않도록 **더 넓은 표현 범위**를 가진 타입으로 자동 변환
- ex) int + float = float

```python
num1 = 100 #int
num2 = 3.123 #float
type(num1), type(num2) # <class 'int'> <class 'float'>
```

```python
num3 = num1 + num2
type(num3) # <class 'float'>, int + float = float : to prevent loss of precision -- the decimal part, the result is float
```

### 4. 형변환 시 주의해야 할 예외와 에러(ValueError)

- 순수 숫자가 아닌 경우 에러: 숫자에 알파벳, 한글, 공백 등 포함 -> int로 바꿀 수 없음
- 소수점이 포함 --> float() 후 int()
- str.isdigit(): **문자열**이 숫자로만 구성되어 있는지 확인, bool

```python
time = "10시"
int(time) # ValueError: invalid literal for int() with base 10: '10시' -- 문자열에 숫자가 아닌 문자가 포함되어 있어 정수로 변환 불가
```

```python
pi = "3.14"
int(pi) # ValueError: invalid literal for int() with base 10: '3.14' -- 문자열에 소수점이 포함되어 있어 정수로 변환 불가
float(pi) # float() 함수를 사용하여 문자열을 실수로 변환 가능
int(float(pi)) # float() 함수를 사용하여 문자열을 실수로 변환 후 int() 함수를 사용하여 정수로 변환 가능
```

```python
time = "10시"
if time.isdigit(): # isdigit() 함수를 사용하여 문자열이 숫자로만 이루어져 있는지 확인 가능
    print(int(time)) # 문자열이 숫자로만 이루어져 있으면 int() 함수를 사용하여 정수로 변환 가능
else:
    print("문자열에 숫자가 아닌 문자가 포함되어 있어 정수로 변환 불가")
```
