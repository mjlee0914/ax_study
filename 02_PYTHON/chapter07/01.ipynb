## 3. try - except - else - finally

예외처리 구문

```
try:
    # 예외가 발생할 가능성이 있는 코드

except:
    # 예외가 발생하면 유입되는 구간, 예외에 대한 대안 처리
    # 대안 로직, 로깅 처리(오류 발생하지 않아도 로깅처리)
    # 여러 개 예외처리 가능

else:
    # 예외가 발생하지 않아야 유입되는 구간

finally:
    # 예외가 발생하든 정상 처리되든 항상 유입되는 구간, 공통적인 처리
    # 자원 정리(메모리 해제)

```


- 모든 예외 클래스는 Exception의 자식 클래스
- 특정 예외를 특정하지 못하는 경우 Exception으로 처리 가능

````python
try:
    num = int(input())
    result = 10 / num

    print(result)
except: 
    print("0이 아닌 정수를 입력하세요.")

else: 
    print("정상처리")
finally:
    print("로깅처리...")
````

````python
try:
    num = int(input())
    result = 10 / num

    print(result)
except ZeroDivisionError:
    print("0이 아닌 정수를 입력하세요.")
````

## 6. raise를 활용한 데이터 유효성 검증

- 예외를 발생시키기 using 예외 객체

```
raise 예외클래스("메시지..")
```

````python
def login(user_id, user_pass):
    if not (user_id == "user01"):
        print("아이디가 일치하지 않습니다.")
        print("아이디를 확인하는 회원 아이디 찾기 페이지로 이동...")
        return

    if not (user_pass == "1234"):
        print("비밀번호가 일치하지 않습니다.")
        return

    print("로그인 처리 로직...")
````

````python
def login(user_id, user_pass):
    if user_id != "user01":
        raise ValueError("아이디가 일치하지 않습니다") # 예외가 발생, 코드 실행 흐름은 중단

    if user_pass != "1234":
        raise ValueError("비밀번호가 일치하지 않습니다")

    print("로그인 처리 로직...")

````

````python
# 회원페이지    
try:
    login("user02", "1234") # 아이디가 일치하지 않음
except ValueError as e:
    print(e)
    print("회원 페이지에서 추가적으로 할 로직...")
````

````python
# 관리자 페이지    
try:
    login("user02", "1234") # 아이디가 일치하지 않음
except ValueError as e:
    print(e)
    print("관리자 페이지에서 추가적으로 할 로직...")
````

## 7. 사용자 정의 예외

- Exception은 모든 예외의 부모클래스
- Exception은 상속 받으면 또는 Exception을 상속받은 자식 클래스를 상속 받으면 그 클래스도 예외 클래스


```
class 예외클래스(Exception):
    def __str__(self):
    return "예외메시지정의"


raise 예외클래스("다른 예외메시지도 정의")
```

````python
class UserIdMismatchError(Exception):

    def __str__(self):
        return f"아이디가 일치하지 않습니다." #예외마다 항상 메시지가 동일할 때 편리

class UserPassMismatchError(Exception):

    def __str__(self):
        return f"비밀번호가 일치하지 않습니다."

class UserIdPassMismatchError(Exception):
    pass
````

````python
def login(user_id, user_pass):
    if user_id != "user01" and user_pass != "1234":
        raise UserIdPassMismatchError("아이디와 비밀번호가 일치하지 않습니다") # raise 예외클래스("상황에 따라 다른 메시지 전달하고 싶을 때")

    if user_id != "user01":
        raise UserIdMismatchError # 예외가 발생, 코드 실행 흐름은 중단

    if user_pass != "1234":
        raise UserPassMismatchError


    print("로그인 처리 로직...")
````

````python
try:
    login("user02", "1234")

except (UserIdMismatchError, UserPassMismatchError, UserIdPassMismatchError) as e:   
    print(e)
````

## 8. traceback

- traceback.print_exc() -- 에러 스택 콘솔 출력
- traceback.format_exc() -- 에러 스택을 문자열로 반환
- traceback.print_stack() -- 에러 없이 현재 호출 흐름 확인


`중요`: 파일위치, 어떤파일, 어떤내용에서 오류가 났는지 확인할 수 있는 점

````python
import traceback

try:
    result = 10 / 0
    print(result)
except:
    print("에러 발생")
    # traceback.print_exc()
    message = traceback.format_exc() # 에러 메시지를 변수로 저장, 나중에 사용 가능
    print("message", message)
````
