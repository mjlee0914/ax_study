## 1. with open()의 자원 자동 해제

`with open() as `
- 시작시 첫 라인에서 \_\_enter\_\_(..) 실행
- 마지막 라인 실행 후 \_\_exit\_\_(..) 실행

*generator* - 값을 필요할 때 하나씩 만들어냄

*with* - 자원을 시작하고 끝날 때 자동 관리

*with open()* 파일 오픈 - 사용 - 자동 닫기

**write**

`with open("file", "w", encoding="UTF-8")`
- overwrite whatever you wrote before

`with open("file", "a", encoding="UTF-8")`
- append next to what you wrote before



```python
class MyResource:
    def __enter__(self):
        print("__enter__ 호출")
        return self

    # with 구문이 종료되었을 때 or 예외가 발생했을 때 호출
    # 예외 발생시 exc_type - 예외 클래스 종류
    #           exc_val - 예외 메시지
    #           exc_tb - TraceBack message
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("__exit__ 호출")
```

```python
with MyResource() as o: # __enter__ 함수 호출
    print(o)
    # 종료시 __exit__ 호출
```

```python
# "w"(write) 파일에서 작성하는 위치(파일 포인터)가 가장 앞으로 이동해서 내용을 대체
with open("test.txt", "w", encoding="UTF-8") as fp:  #UTF-8로 한글을 온전하게 입력
    fp.write("Hello!!")
    fp.write("hey!")
```

```python
# "a"(append) 파일 포인터를 이동하지 않고 누적해서 작성 
with open("test.txt", "a", encoding="UTF-8") as fp:
    fp.write("\n안녕하세용")
```

## 2. 텍스트 파일 읽기 함수 (read, readlline, readlines)

**read**

`with open("file", "r", encoding="UTF-8")`

- `read()`: 한꺼번에 전체 텍스트를 읽어 옴
- `readline()`: 한 줄 읽어오기, 한 줄씩 읽어서 메모리에 적재하므로 메모리 소비 적음
- `realines()`: 라인별로 읽어서 리스트로 반환, 각 라인은 하나의 요소, 한꺼번에 메모리에 올림 -> 메모리 소리 많음, 큰 파일에 적합하지 않음

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    text = fp.read()
    print(text)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    line = fp.readline()
    print(line)
    line = fp.readline()
    print(line)    
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    while(text := fp.readline()): # <코끼리 연산자>a := b 는 b를 a에 대입해줌, 값 없으면 False 있으면 True
        print(text)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    lines = fp.readlines()
    print(lines)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    for line in fp:
        print(line)
```

## 3. CSV 형식


Comma Seperated Value (CSV)
    - writer(파일포인터)
        - writerow(iterable)
    - reader(파일포인터)

```python
import csv

with open("member.csv", "w", encoding="UTF-8", newline="") as fp: #파일에 이미지 등 있을 때 wb 사용 (binary), newline=""로 empty line delete
    writer = csv.writer(fp)
    writer.writerow(["이름", "나이", "전화번호"])
    writer.writerow(["김철수", 20, "010-1000-1000"])        #writerows 여러줄 

```

```python
with open("member.csv", "r", encoding="UTF-8") as fp:
    reader = csv.reader(fp)
    members = list(reader)
    print(members)
```

## 4. JSON 형식

- 파이썬 객체 -> 전송 가능한 JSON 형식으로 변환 - 직렬화 (Serialize)
- JSON 문자열(전송 받은 문자열) -> 파이썬 객체 - 역직렬화 (Deserialize)
- `json.dump()`
- `json.load()`

```python
import json

# python dict -> JSON file 직렬화
user = {
    "name": "김영희",
    "age": 20,
    "mobile": "010-0000-0000"
}

with open("user.json", "w", encoding="UTF-8") as fp:
    json.dump(user, fp, indent=4, ensure_ascii=False)


# json file -> python object(dict) 역직렬화
with open("user.json", "r", encoding="UTF-8") as fp:
    user2 = json.load(fp)
    print(user2)
    print(type(user2))
```
