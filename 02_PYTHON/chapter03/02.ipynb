## 1. 딕셔너리(Dictionary) 자료구조

- key - value 쌍 구조
- key : value를 찾기 위한 키워드, 중복 x, 리스트는 사용 불가, 튜플은 가능(불변)
- value: key로 찾는 값, 중복 o
- {"key":val, "key":val, ...}

- Create: 없는 키 이름으로 값을 대입하여 추가
    - `dict['key'] = val`

- Read: dict['key'] - **키 없으면 KeyError 발생**
    - `get("key")`: 키가 없으면 **None** (dict["key"]와 차이점)
    - `get('key', '기본값')`: 키 없으면 기본값 출력
    - `keys()`: 모든 키를 담은 view, 읽기 전용
    - `values()`: 모든 값을 담은 view, 읽기 전용
    - `items(`): (key, value) 목록을 담은 view

- Update: 이미 있는 키값에 대입하여 수정
    - `update({'key': val, 'key': val, ...})` : dict가 안에 들어감
    

- Delete: 
    - `del items['key']` : key-value 쌍 삭제. 변수에 할당 불가(`del`은 statement)

        **`.remove`와 차이?**   
            - dictionary에는 `.remove` 존재 x
            - `list.remove("value")`: value 삭제
            - `del items['key']`: key 기준으로 key-value pair 삭제
    
    - `pop()`: key-value pair를 삭제하면서 **삭제된 value 반환** . 변수에 반환된 value 할당 가능
    - `pop('key', default value)`
    - `clear()`: 전체 비우기

**hash vs id**
- `hash('key')`: dict의 hash table에서 key 저장/검색 위치 결정하는 hash value를 return
    - hash collision: 다른 두 객체의 hash 값이 같은 경우 
- `id('key')`: key 객체의 identity, 두 변수가 같은 객체를 참조하는지 확인할때 등 사용

다른 키-값 쌍 자료형 -> 딕셔너리로 변환
- `dict()`

```python
user = {
    "email": "test@test.org", 
    "password": 1234,
    "name": "김철수"
    }

# create
user['mobile'] = "010-0000-0000"
print(user)

# read
print(user.keys())
print(user.values())
print(user.items())
print(user['mobile']) # return value, if not KeyError

print(user.get('address')) # return None
address = user.get('address')
print(type(address)) # return class 'NoneType'
print(user.get('address', "없어")) # if None -> "없어" 출력

# update
user.update({'mobile': "010-1111-2222", 'password': 1235})
print(user)

# delete
del user['mobile']
print(user)

# hash -- key를 hash function으로 변환한 정수값으로, 해시 테이블에서 저장/검색 위치를 결정하는 데 사용(value값 찾기 목적)
print(hash("name"))
# key id 찾기
print(id("name")) 
# dict의 모든 key id 찾기
for key in user:
    print(key, id(key))
```

## 3. 집합(Set) 자료구조

- 중복 x
- 순서 x -> 위치 계산 x. index 사용 불가
- {값1, 값2, 값3, ...}


#### 집합 연산
- 합집합: `a | b` -> `a.union(b)`
- 교집합: `a & b` -> `a.intersection(b)`
- 차집합: `a - b` -> `a.difference(b)`
- 대칭차집합: `a ^ b` -> `a.symmetric_difference(b)`



#### 집합 자료로 형변환
`set(리스트 또는 튜플)`


```python
a = {10, 20, 10, 30, 30}
print(a) # 중복 x
print(a[0]) # TypeError
```

```python
a = {10, 20, 30, 40}
b = {30, 40, 50, 60}

union = a.union(b)
print("합집합: ", union)
print(a | b)

intersec = a.intersection(b)
print("교집합: ", intersec)
print(a & b)

diff = a.difference(b)
print("차집합: ", diff)
print(a - b)

sym_diff = a.symmetric_difference(b)
print("대칭차집합: ", sym_diff)
print(a ^ b)
```
