## 3. [데이터 표준 변환] json

**JSON- JavaScript Object Notation 자바스크립트 객체 표기법**

- 용도: 다양한 언어 플랫폼간 공통 의사소통을 할 수 있도록 함 <언어에 관계없이 통일된 데이터를 주고받을 수 있도록 일정한 패턴을 지닌 문자열을 생성>

- 자바스크립트의 객체를 정의하는 형식을 빌려온 것

- 자바스크립트가 아닌 *문자열*

- `{"이름": "값", "이름": "값", ...}`


- `json.dumps(obj)`: 파이썬 객체(리스트, 딕셔너리 등)를 JSON 문자열로 반환
- `json.loads(s)`: JSON 문자열 -> 파이썬 객체



```python
import json

# 파이썬 객체 -> JSON 문자열
user = {
    "name": "김철수kim철수", # UNICODE 16진수 (json.dumps(obj, ensure_ascii=False))하면 원본 문자열 나옴
    "age": 20
}

json_string = json.dumps(user, ensure_ascii=False)
print("json_string:", json_string, "type:", type(json_string))
```

```python
# JSON 문자열 -> 파이썬 객체

nums_string = "[1, 2, 3, 4, 5]"
user_string = '{"name": "김철수", "age": 20}'

nums = json.loads(nums_string)
user2 = json.loads(user_string)

print("nums: ", nums, "type: ", type(nums))
print("user2: ", user2, "type: ", type(user2))
```

## 4. [무작위 난수 추출] random

- `random.randint(a, b)`: a 이상 b *이하* int 반환
- `random.random()`: 0.0 이상 ~ 1.0 *미만* float 반환
- `random.choice(seq)` : sequence 무작위 요소 반환, 중복 가능
- `random.shuffle(seq)`: 원본 리스트 무작위 뒤섞기
- `random.sample(seq, k)` : seqeuence 무작위 요소 k개 만큼 반환, 중복 없음
