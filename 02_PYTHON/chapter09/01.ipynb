## 2. all()과 any()

- `all(iterable)`: **모든** 요소가 True로 인식될 때 -> `True`   예외) 빈 리스트([]) = `True`
- `any(iterbale)`: 요소 중 **하나라도** True 인식이 되면 -> `True`

```python
print(all([True, 1, " "]))
print(all([False, False]))
print(all([]))
print(all([False, 1, 0]))
```

## 3. len(), sum(), round()

`len()`: 문자열-문자열 길이, 리스트/튜플 -> 요소의 개수

`sum(iterable)`: 요소의 합계 (조건: 요소는 숫자로만 구성, string-> 오류)

`round(self, num)`: 반올림 소수점뒤 num까지
- Math.ceil(..): 올림
- Math.floor(..): 버림



```python
import math

n = 3.134567

print(round(n, 3))
print(math.floor(n))
print(math.ceil(6.14))
```

## 4. mean()과 median()   

- statistics module에 포함된 표준 라이브러리
- `mean(iterable)` : 평균 구하기 iterbles should be numeric
- `median(iterable)` : 중앙값 iterbles should be numeric

```python
import statistics

nums = [10, 15, 20, 8, 7, 1000000]

print("평균: ", statistics.mean(nums))
print("중앙값: ", statistics.median(nums))
```

## 5. map()과 filter()

- `map(function, iterable)`: 변환작업
- `filter(조건function, iterable)`: 판별 -> 정제
--> 리스트 컴프리헨션으로 대체 가능 

```python
nums = [*range(1, 11)]

nums2 = [x**2 for x in nums]

nums3 = [x for x in nums if x % 2 == 1]
print(nums3)

nums4 = list(map(lambda x: x*2, nums))
print(nums4)
```

## 6. enumerate()

`enumerate(iterable, start=num)`
- for문에서 item print --> tuple 형태로 출력, unpack tuple!

```python
fruits = ['사과', '오렌지', '망고', '멜론']

for fruit in fruits:
    print(fruit)


for idx, fruit in enumerate(fruits):
    print(f"{idx}번 {fruit}")

for item in enumerate(fruits):
    print(item) # tuple 형태


for item in enumerate(fruits):
    idx = item[0]
    fruit = item[1]
    print(idx, fruit)
```

## 7. zip()

`zip(*iterables, strict=False)`


참고 - intertools 모듈 zip_longest

```python
first_name = ["Hayley", "Saurav"]
last_name = ["Williams", "Krishna"]

full_name = zip(first_name, last_name)  # --> zip(several iterables) = > zipped tuple
print(full_name) # tuple

for index, (firstname, lastname) in enumerate(full_name, start=0):
    print(index, firstname, lastname)
```

## 9. min(), max()

- `min(iterable)`: 최소값 / 오름차순 정렬 후 가장 처음 요소
- `max(iterable)`: 최대값 / 오름차순 정렬 후 가장 마지막 요소

```python
nums = [*range(1, 50)]

min_nums = min(nums)
max_nums = max(nums)

print(min_nums, max_nums)
```

```python
fruits = ['사과', '오렌지', '바나나', '멜론']

print(min(fruits), max(fruits))
```
