```python
# 문제 3-1: 무제한 과목 점수 합계 및 평균 산출
def get_total_and_average(*arg):
    if len(arg) == 0:
        return 0
    else: 
        total = sum(arg)
        avg = total / len(arg)
        return total, avg
       

print(get_total_and_average())
```

#### chap 05-3 클로저와 데코레이터  

```python
# 문제 1
def message_store():
    storage = ""

    def add_message(message):    
        nonlocal storage
        storage += "\n" + message
        return storage

    return add_message 

my_storage = message_store()

print(my_storage("새로운 알림이 도착했습니다"))
print(my_storage("출석 체크가 완료되었습니다"))
```

```python
# 문제 2

def deco_border(callback):
    def wrapper():
    
        print("=" * 18)
        callback()
        print("=" * 18)

    return wrapper


@deco_border
def show_welcome():
    print("환영합니다 스파르타님")

show_welcome()
```

```python
# 문제 3
def auto_tracker(callback):
    def wrapper(*arg, **kwargs):
        print(f"원본 함수 <{callback.__name__}> 실행 시작!")

        print(f"[위치 인수: {arg} | 가변 인수: {kwargs}] 입력!")

        result = callback(*arg, **kwargs)

        print(f"<{callback.__name__}> 연산 완료!")
        
        return result

    return wrapper

@auto_tracker
def add_numbers(a, b):
    return a + b

add_numbers(10, b = 20)
```
