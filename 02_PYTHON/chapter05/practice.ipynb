```python
# 문제 1-1: 대여 완료 안내 메시지 출력
def rent_book(user_name, book_title):
    print(f"[회원 이름: {user_name} | 책 제목: {book_title}] 대여 완료")

rent_book("홍길동", "파이썬 안내서")
rent_book(book_title="데이터 구조", user_name="이영희")
```

```python
# 문제 2-1: 배송비 자동 계산 및 다중 데이터 반환
def calculate_total(price, delivery_fee = 3000):
    return price + delivery_fee, "결제가 완료되었습니다"

calculate_total(25000)
total, message = calculate_total(25000)
print(total, message)
```

```python
# 문제 3-1: 무제한 과목 점수 합계 및 평균 산출
def get_total_and_average(*arg):
    if len(arg) == 0:
        return 0
    else: 
        total = sum(arg)
        avg = total / len(arg)
        return total, avg
       

print(get_total_and_average(90, 85, 95)) 
print(get_total_and_average(70, 80, 90, 100, 65))
```
