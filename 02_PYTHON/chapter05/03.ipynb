## 2. Closure

- 고차 함수 (Higher Order Function)

```python
def outer():
    num1 = 10

    def inner():

        num2 = 20
        print(num1 + num2)

    return inner


inner = outer() # num1은 inner에 의해서 사용되므로 heap 영역에 생성
inner()
```

```python
def add(num1):
    def _add(num2):
        def __add(num3):
            
            return num1 + num2 + num3
        
        return __add
    
    return _add

result1 = add(10)(20)(30)
print(result1)
```

```python
add10 = add(10)
result2 = add10(20)(30)
print(result2)
add10_20 = add10(20)
result3 = add10_20(30)
print(result3)
```

## 3. Decorator

- 디자인 패턴 중 한가지

**factorial**
- factorial n! = 1부터 n까지 모든 양의 정수 곱하기 ex) n * (n-1) * (n-2) * ... * 2 * 1


```python
def factorial1(num):
    result = 1

    for i in range(1, num + 1):
        result *= i

    return result

print(factorial1(10))
```

```python
import time

start = time.time() # 공통 기능

result = factorial1(10) # 핵심 기능
print(result)

end = time.time() # 공통 기능
print("걸린 시간: ", end - start)
```

```python
# 같은 함수 내에서 같은 함수를 반복해서 호출하는 방법 -> recursive (재귀함수).
# factorial1보다 성능이 좋지 않음.

def factorial2(num):

    if (num < 1):
        return 1
    
    return num * factorial2(num - 1)


start = time.time()

result = factorial2(10)
print(result)

end = time.time()
print("걸린 시간: ", end - start)

```

```python
# decorator pattern

import time


def time_checker(callback, num):
    
    start = time.time() # 공통 기능 

    result = callback(num) # 핵심 기능 - 팩토리얼 연산 함수 (factorial1, factorial2) --> decorator. 
    
    end = time.time() # 공통 기능

    print("걸린 시간:", end - start )

    return result

time_checker(factorial1, 10)
```

```python
# time_checker2 = decorator
import time

def time_checker2(callback):
    def wrapper(*args, **kwargs): # 관례적으로 포장된 함수 wrapper로 지음, 핵심 기능에 해당, *args/**kwargs로 정의하여 나중에 넣을 핵심 기능에 어떤 인자가 오더라도 핸들 가능
        start = time.time() # 공통 기능

        result = callback(*args, **kwargs) # 핵심 기능

        end = time.time() # 공통 기능

        print("걸린 시간:", end - start)

        return result

    return wrapper
```

```python
@time_checker2 ##알아서 time_checker2 인수로 들어감
def factorial2(num):

    if (num < 1):
        return 1
    
    return num * factorial2(num - 1)
```

```python
factorial2(10)
```

```python
@time_checker2
def square(num):
    return num * num

square(8)
```

```python
@time_checker2
def add(num1, num2):
    return num1 + num2

add(1, 2)
```

```python
a = [1, 2]

def minus(num1, num2):
    return num1 - num2

minus(*a) # *를 사용해 리스트/튜플을 펼쳐서 전개시킴
```
