## 1. Iterable

- 반복 가능한 객체
- `\_\_iter\_\_(..)` 가 정의되어 있다면 반복 가능 -> iterator 만들어냄
- `\_\_next\_\_(..)` 인덱스를 한칸씩 이동하면서 요소를 꺼내오는 기능을 정의
- 이터레이터(반복자)

```python
nums = [10, 20, 30]

nums.__iter__

for num in nums:
    print(num)
```

## 2. Iterator

iterator(반복 가능한 객체) 만들기
1. iterator 객체 생성(`__iter__` 호출(iter(...)))
2. 순서를 한 칸씩 이동하면서 요소 조회(`__next__` 호출(next(...)))

* 외부에서 iter(obj) 사용하면 내부적으로 obj._\_iter_\_()가 호출됨

**enumerate()**
- iterator를 만들어 주는 함수
- 반복하면서 'index num + value' 같이 줌

```python
fruits = ['Apple', 'Orange', 'Melon', 'Banana']

# 1. iterator 만들기
it = iter(fruits) # --> __iter__(..) 호출

# 2. next() 호출, 호출할 때마다 다음 인덱스로 이동하면서 요소 조회 가능
print(next(it)) # 내부적으로 위치를 기록, 0번째 인덱스부터
print(next(it)) # 다음 번호 기록해서 1번째 인덱스의 value 호출 
```

## 3. Generator

- 함수의 실행을 정지하고 원할 때 다시 재개하는 형태 - lazy evaluation
- `yield` 키워드 만날 때마다 함수 실행 중지
- next()로 호출하면 함수 재개 
- 꺼내올 때마다 힙 영역에 해당 값을 할당 -> 메모리 절약, 큰 파일 처리할 때 사용

```python
def generate_nums():

    print("1번째")
    yield 10 # 함수 실행 중지 (종료x)

    print("2번째")
    yield 20

    print("3번째")
    yield 30

gen = generate_nums()

print(gen)
```

```python
print(next(gen)) #호출 시마다 yield 중지 -> next() 호출 -> yield 중지
                 #호출 시마다 힙 영역에 해당 값을 할당
```

```python
nums = [10, 20, 30] #처음부터 힙 영역에 10, 20, 30이 할당
```

**동기 vs 비동기**

동기 synchronous
- 순서대로 실행하고 결과도 순서대로 도출 (이전 작업 종료 -> 다음 작업 시작)
- a 작업 시작 -> a 작업 종료 -> b 작업 시작 -> b 작업 종료 -> c 작업 시작 ...

비동기 asynchronous
- 순서대로 작업을 시작하지만 결과는 먼저 처리된 순서대로 나오는 방식
- 작업 완료시 '이벤트 발생'


# 5. 제너레이터 표현식

- 참고: 리스트 컴프리헨션 -> 생성 결과가 처음부터 메모리에 적재
    * 리스트 컴프리헨션 : `[표현식 for 변수 in 반복가능한객체]`
- 리스트 대신 **튜플**로 리스트 컴프리헨션과 동일하게 사용하면 된다. -> 메모리 소비가 적다.
- 대용량 데이터 처리
- `yield`가 포함된 구조

```python
gen = ( i for i in range(1, 101) if i % 2 == 0) # 아직 메모리에 적재되지 않음
gen
```

```python
next(gen) #next()할 때마다 generator 값 메모리에 적재
```

```python
for num in gen:
    print(num) # next(gen) 으로 출력한 값부터 세서 출력
```
