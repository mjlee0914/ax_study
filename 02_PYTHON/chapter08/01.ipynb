## 2. 클래스(Class)와 인스턴스(Instance)

클래스
- 객체를 만들기 위해서 정의된 명세서
- 객체가 되면 갖추어야 할 변수, 함수 정의

```python
class Bus:
    # 속성 정의
    def __init__(self):
        self.count = 0 # 승객 수, 초기화: 변수가 되었을 때 최초로 값을 대입 "객체 생성 직후"
        self.passengers = [] # 승객 담을 공간

    # 행위 정의
    def ride(self, passenger): 
        self.count += 1
        self.passengers.append(passenger)
        print("탑승")


    # 다른 class와 서로 상호작용 예시
    def off(self, passenger): 
        if self.count > 1:
            self.count -= 1
            self.passengers.remove(passenger)

        print("하차")

    def show_count(self):
        print("승객 수: ", self.count)


    def print_self(self):
        print(id(self))

    def __str__(self):
        return f"승객 수: {self.count} | 승객: {self.passengers}"

    def __repr__(self):
        return "REPR"


class Passenger: 
    def __init__(self, name, age): # (..., name, age) => 객체를 생성하기 위해서 매개변수로 정의된 지역변수
        self.name = name  # self.name, self.age => 객체가 생성이 되면 만들어지는 인스턴스 변수
        self.age = age 

    def __str__(self):
        return f"이름: {self.name}, 나이: {self.age}"

    def __repr__(self):
        return f"Passenger(name='{self.name}', age={self.age})" # convention
        
```

```python
bus1 = Bus() # 객체, 클래스 -> 객체(인스턴스 = 실체)

# 마침표(.) 연산자를 통해서 객체의 (인스턴스)변수, 메서드에 접근 가능 

print("탑승객 수: ", bus1.count)
bus1.ride()

print("bus1 주소:", id(bus1))
bus1.print_self()

bus1.ride()
bus1.off()

bus1.show_count()
```

```python
bus2 = Bus()

bus1 is bus2 # False, 같은 클래스로부터 생성된 객체는 서로 다른 객체
```

```python
bus = Bus()
p1 = Passenger("김철수", 20) 
p2 = Passenger("김영희", 18)

bus.ride(p1)
bus.ride(p2)
bus.show_count()
```

## 6. \_\_str\_\_과 \_\_repr\_\_

`__str__`
- print(객체 참조변수) 시 자동으로 호출되는 메서드
- 보통 객체 내 인스턴스 변수의 내용을 출력하는 경우가 많다



`__repr__`
- print(객체 참조변수) 시 자동으로 호출(\_\_str\_\_과 비슷)
- \_\_str\_\_과 동시 출력시 str이 우선순위 높음 
- 로깅 목적 (기록을 위한 목적, 디버깅용)
- 클래스명(매개변수='값', ...) 형식으로 주로 정의됨 -> `eval(문자열)` 을 실행햇을 때 객체가 만들어지는 형태
- `repr(객체 참조변수)`로 호출

- `eval(문자열)`: 문자열을 파이썬 코드로 해석해서 실행해줌 / 보안상 위험할 수 있으므로 사용에 유의



```python
bus = Bus()

p1 = Passenger("김철수", 20) 
p2 = Passenger("김영희", 18)

bus.ride(p1)
bus.ride(p2)

print(bus)
```

```python
p1 = Passenger("김철수", 20)

print(p1) # __str__호출
print(repr(p1)) # __repr__ 호출
```
