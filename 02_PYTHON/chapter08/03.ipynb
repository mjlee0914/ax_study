## Class, Call Stack, and Heap


Call Stack
┌──────────────────────┐
│ Frame                │
│ user1 ───────┐       │
│ user2 ────┐  │       │
└───────────│──│───────┘
            │  │
            ↓  ↓
Heap
┌───────────┐ ┌───────────┐
│ User #1   │ │ User #2   │
│ name=Kim  │ │ name=Lee  │
└───────────┘ └───────────┘

- user1 and user2 are references (names) in the Frame.
- user1 → User #1, user2 → User #2.
- The actual User objects live in the Heap.
- The Call Stack stores the current function's Frame.
- The Class User is also an object in the Heap.

# 2. @classmethod, @staticmethod

**instance method**
- 클래스 내부에서 정의
- 클래스를 통해 생성된 객체(인스턴스)에서 호출
- 첫 번째 매개변수 `self`, 호출 시 인스턴스를 전달

```python
class Pokemon:
    def __init__(self, name, type): # 인스턴스 생성 시 호출. name, type 초기화
        self.name = name
        self.type = type

    def hungry(self): # instance method
        print(f"{self.name}이(가) 배가 고픕니다")

    def info(self): # instance method 
        print(f"포켓몬: {self.name}")
        print(f"타입: {self.type}")
```

```python
pikachu = Pokemon("Pikachu", "Electric") # instance from class 'Pokemon', having 2 attributes 'name', 'type'
```

**@class method**
- `@classmethod` decorator를 사용하여 정의
- 클래스 변수 조작 가능
- 새로운 인스턴스 생성 가능
- 첫 번째 인자로 cls를 받음 -> 클래스 자체를 전달 == 클래스 전체에 관련된 작업

```python
class Student:
    school = "Old school" # 클래스 변수

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school
```

```python
print(Student.school)
Student.change_school("New school")
print(Student.school)
```

**@static method**
- `@staticmethod` decorator를 사용하여 정의
- 첫 번째 인자로 self, cls 받지 않음
- 클래스 내부 데이터 접근 불가 (클래스/인스턴스 관련 작업 안함)
- 인스턴스 속성, 인스턴스 메서드 필요 없을 때 사용
- 메서드 실행이 외부 상태 영향 미치지 않는 순수 함수 만들 때 사용

```python
class Student:

    @staticmethod
    def add(a, b):
        return a + b
```

```python
Student.add(5, 5)
```

```python
class Calculator: 
    count = 0 # 클래스 변수

    # cls -> Calculator 클래스 객체 참조
    # cls -> 클래스 변수, 다른 클래스 메서드를 접근하기 위한 목적
    @classmethod # @classmethod decorator
    def add(cls, num1, num2):
        cls.count += 1
        print("호출 횟수:", cls.count)
        return num1 + num2

    @staticmethod
    def minus(num1, num2):
        return num1 - num2
```

```python
Calculator.add(10, 20)
Calculator.minus(10, 20)
```

## real-life exmplae of classmethod/staticmethod

```python
class Product:
    tax_rate = 0.10

    def __init__(self, name, price):
        self.name = name
        self.price = price

    # 1. Instance method 
    # product instance 
    def price_with_tax(self):
        return self.price * (1 + Product.tax_rate)

    # 2. Class method
    @classmethod
    def change_tax_rate(cls, new_rate):
        cls.tax_rate = new_rate

    # 3. Static method
    @staticmethod
    def is_valid_price(price):
        return price >= 0
```

```python
apple = Product("Apple", 2)
banana = Product("Banana", 3)

apple.price_with_tax()
```

```python
Product.change_tax_rate(0.20) #classmethod
```

```python
Product.is_valid_price(-5) #staticmethod
```
