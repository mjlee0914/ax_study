## 1. 상속 Inheritance

- 클래스 기반의 상속(상속에 대한 정의 -> 객체 생성(상속 관계 성립))


```

class 자식클래스(부모클래스): # 자식 class가 부모 class를 상속받음
        ...

```


- super: 부모 객체를 가르키는 참조 변수
    - 참고) self: 자식 객체(현재 객체)의 참조 변수
    - super() : 부모쪽 객체 생성, self 생략


- 부모클래스
    - 객체에 대한 일반적인 정의들 (상위 개념)

- 자식클래스
    - 객체에 대한 구체적인 정의들 (하위 개념)

````python
class Animal:
    def __init__(self, kind):
        self.kind = kind # 동물 종류
    
    def move(self):
        print("동물이 움직인다.")



class Dog(Animal):
    def __init__(self):
       super().__init__("강아지") # 부모쪽 인스턴스 변수인 kind에 초기화

    def move(self): # method overriding
        super().move() # 부모쪽 객체의 move(), super().부모쪽 메서드() 일 경우 self X
        print(self.kind + " 걸음으로 움직인다.")    



class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print(self.kind + " 걸음으로 움직인다.")
````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()

````

````python
print(dog.kind, cat.kind)
````

## 2. 메서드 오버라이딩(Overriding)

- 메서드 재정의
- 부모쪽 클래스와 동일한 이름으로 정의하여 기능을 수정, 재정의
- 자식쪽 클래스에 정의된 동일한 이름의 함수가 호출 우선순위를 가짐
- 자식쪽에서 부모쪽 동일한 이름의 메서드를 호출해야하는 경우는 self로 접근 못함
    - super().부모쪽 메서드(..)

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
````

````python
class Human:
    def move(self):
        print("두 발로 직립보행한다.")
````

````python
human = Human()
human.move()
````

> 객체의 출처(클래스 자료형) 확인하는 함수 - isinstance(참조 변수 / 클래스 / 자료형)

`isinstance(객체, 클래스or자료형)`

````python
print("human은 Animal 클래스로부터 생성되었나?", isinstance(human, Animal))
print("human은 Human 클래스로부터 생성되었나?", isinstance(human, Human))

print("dog은 Animal 클래스로부터 생성되었나?", isinstance(dog, Animal))
print("cat은 Animal 클래스로부터 생성되었나?", isinstance(cat, Animal))
````

````python
s = "ABC"
#s.lower()

# 문자열인지 체크
isinstance(s, str)
````

> 다중 상속 - 여러 부모 클래스를 상속 관계로 정의 가능

- 다중 상속시 같은 메서드가 있을 때 호출 우선순위 존재 (Method Resolution Order)
- `자식클래스.__mro__`로 우선순위 확인
- 자식 클래스 메서드 > 상속 관계에서 왼쪽에 있는 클래스 > 오른쪽 클래스


````python
class ParentA:
    def a(self):
        print("ParentA - a()")

    def b(self):
        print("ParentA- b()")

class ParentB:
    def a(self):
        print("ParentB - a()")

    def c(self):
        print("ParentB - c()")



# class Child(ParentA, ParentB): # 다중 상속 by ,
#     pass

class Child(ParentB, ParentA):
    pass

````

````python
c = Child()
c.a() # 같은 이름의 메서드가 호출됨 -> 자식 클래스에서 부모 클래스가 정의된 순서대로 호출됨
c.b()
c.c()
````

````python
Child.__mro__ # Method Resolution Order: 메서드 우선순위 확인
# mro 호출 시 마지막 object = 가장 상위 클래스, 사실상 모든 클래스엔 object라는 최상위 클래스의 매개변수 존재
````
