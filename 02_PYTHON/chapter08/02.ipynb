## 1. 상속 Inheritance

- 클래스 기반의 상속(상속에 대한 정의 -> 객체 생성(상속 관계 성립))


```

class 자식클래스(부모클래스): # 자식 class가 부모 class를 상속받음
        ...

```


- super: 부모 객체를 가르키는 참조 변수
    - 참고) self: 자식 객체(현재 객체)의 참조 변수
    - super() : 부모쪽 객체 생성, self 생략


- 부모클래스
    - 객체에 대한 일반적인 정의들 (상위 개념)

- 자식클래스
    - 객체에 대한 구체적인 정의들 (하위 개념)

````python
class Animal:
    def __init__(self, kind):
        self.kind = kind # 동물 종류
    
    def move(self):
        print("동물이 움직인다.")



class Dog(Animal):
    def __init__(self):
       super().__init__("강아지") # 부모쪽 인스턴스 변수인 kind에 초기화

    def move(self): # method overriding
        super().move() # 부모쪽 객체의 move(), super().부모쪽 메서드() 일 경우 self X
        print(self.kind + " 걸음으로 움직인다.")    



class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print(self.kind + " 걸음으로 움직인다.")
````

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()

````

````python
print(dog.kind, cat.kind)
````

## 2. 메서드 오버라이딩(Overriding)

- 메서드 재정의

- 자식쪽에서 부모쪽 동일한 이름의 메서드를 호출해야하는 경우는 self로 접근 못함
    - super().부모쪽 메서드(..)

````python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
````

````python
class Human:
    def move(self):
        print("두 발로 직립보행한다.")
````

````python
human = Human()
human.move()
````

> 다중 상속 - 여러 부모 클래스를 상속 관계로 정의 가능

- 다중 상속시 같은 메서드가 있을 때 호출 우선순위 존재 (Method Resolution Order)
- `자식클래스.__mro__`로 우선순위 확인
- 자식 클래스 메서드 > 상속 관계에서 왼쪽에 있는 클래스 > 오른쪽 클래스


````python
class ParentA:
    def a(self):
        print("ParentA - a()")

    def b(self):
        print("ParentA- b()")

class ParentB:
    def a(self):
        print("ParentB - a()")

    def c(self):
        print("ParentB - c()")



# class Child(ParentA, ParentB): # 다중 상속 by ,
#     pass

class Child(ParentB, ParentA):
    pass

````

````python
c = Child()
c.a() # 같은 이름의 메서드가 호출됨 -> 자식 클래스에서 부모 클래스가 정의된 순서대로 호출됨
c.b()
c.c()
````

````python
Child.__mro__ # Method Resolution Order: 메서드 우선순위 확인
# mro 호출 시 마지막 object = 가장 상위 클래스, 사실상 모든 클래스엔 object라는 최상위 클래스의 매개변수 존재
````

## 4. 정보 은닉과 캡슐화(Encapsulization)

````python
class Schedule:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    def __str__(self):
        return f"year={self.year}, month={self.month}, day={self.day}"
````

````python
date = Schedule(2026, 8, 18)
print(date)
````

````python
# 외부 참조를 통한 instance 값 변경/할당 -> 대입하면 값이 바로 바뀜, 의도치 않은 값이 할당될 수 있어 오류 발생 가능성 높음. "통제 불가능"
# instance 변수에 직접 대입이 가능하므로 값에 대한 통제 불가 --> 통제 불가능하면 직접 접근을 막아야 함.
# --> 인스턴스 변수를 외부에서 접근하지 못하게 막는다

date.year = 2000
date.month = 2
date.day = 28
print(date)
````

## 5. 파이썬 프라이빗(Private) 변수와 이름 변형 규칙(Name Mangling)

private 변수

`__변수명`: 클래스 내부에서만 접근 가능한 인스턴스 변수가 된다

````python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day

    def __str__(self):
        return F"year={self.__year} | month={self.__month} | day={self.__day}"
````

````python
s1 = Schedule(2026, 8, 18)
print(s1)
````

````python
# 접근 불가 -> 값 변하지 않음. 정보 은닉.
s1.__year = 2026
s1.__month = 2
s1.__day = 31

print(s1)
````

## 6. Getter와 Setter Method

#### 통제

````python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day


    # getter:  값을 안전하게 꺼내서 전달
    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month

    def get_day(self):
        return self.__day

    # setter: 올바른 값인지 조건 검증 후 변경
    def set_year(self, year):
        self.__year = year

    def set_month(self, month):
        self.__month = month

    def set_day(self, day):
        if self.__month == 2 and day > 28: # 값에 대한 통제가 가능
            print("2월은 day > 28 의 숫자만 가능합니다.")

        else: self.__day = day


    def __str__(self):
        return F"year={self.__year} | month={self.__month} | day={self.__day}"
````

````python
s1 = Schedule(2000, 8, 18)
s1.set_month(2)
s1.set_day(31)

print(s1)
````
