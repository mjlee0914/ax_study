```python
import numpy as np

"""
np.linalg.norm(): 유클리디언 정규화
"""

def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

def euclidean_similarity(v1, v2):
    return np.linalg.norm(np.array(v1) - np.array(v2))
```

```python
import os
from dotenv import load_dotenv
from langchain_openai import OpenAIEmbeddings

load_dotenv()  # .env 파일 내용을 환경변수로 불러오기

embedding_model = OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=os.getenv("OPENAI_API_KEY")  # .env에 적힌 이름과 똑같이
)
```

```python
fruit1 = "puppy"
fruit2 = "dog"
fruit3 = "Apple"
embedding_fruit1 = embedding_model.embed_query(fruit1)
embedding_fruit2 = embedding_model.embed_query(fruit2)
embedding_fruit3 = embedding_model.embed_query(fruit3)
```

```python
embedding_fruit1[:5], len(embedding_fruit1)
```

```python
# 코사인 유사도
cosine_similarity(embedding_fruit1, embedding_fruit2)
```

```python
cosine_similarity(embedding_fruit1, embedding_fruit3)
```

```python
# 유클리디언 유사도
euclidean_similarity(embedding_fruit1, embedding_fruit2)
```
