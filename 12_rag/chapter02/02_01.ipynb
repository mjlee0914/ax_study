```python
from langchain_community.document_loaders import PyMuPDFLoader

pdf_path = "academic_rules_2026.pdf"
pdf_loder = PyMuPDFLoader(file_path=pdf_path)

pdf_document = pdf_loder.load()

print(f"페이지 수: {len(pdf_document)}")

for i, doc in enumerate(pdf_document, 1):
    print(f"[{i}번째] {doc}")
    print("-" * 50)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

rec_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
)

pdf_docs = rec_splitter.split_documents(documents=pdf_document)
for i, doc in enumerate(pdf_docs, 1):
    print(f"[{i}번째] {doc}")
    print("-" * 50)
```

```python
from langchain_community.document_loaders import WebBaseLoader

target_url = "https://www.gutenberg.org/cache/epub/84/pg84-images.html"

web_loader = WebBaseLoader(web_path=target_url)

web_documents = web_loader.load()
text = web_documents[0].page_content

print(f"전체 길이: {len(text):,}자")
print("Chapter 24" in text)   # True면 마지막 장까지 들어온 것
# print(text[-500:])    

print("=== [웹 문서 로딩 결과] ===")
print(f"문서 제목 및 출처 메타데이터: {web_documents[0].metadata}")
print(f"\n추출된 본문 텍스트 (일부): {text.strip()[1000:10000]}")



```
